## 4.3. Implementación de BFS

BFS realiza una búsqueda en amplitud utilizando una cola FIFO.

Explora primero los nodos más cercanos al estado inicial y permite encontrar el camino con menor cantidad de pasos.

```python
def bfs(grafo, inicio, objetivo):
    """
    Busca el camino mas corto entre 'inicio' y 'objetivo' usando BFS.
    """
    cola = [[inicio]] # cola de caminos, empieza con el estado inicial
    visitados = set()

    while cola:
        camino = cola.pop(0) # sacar el primer camino de la cola (FIFO)
        nodo_actual = camino[-1]

        if nodo_actual == objetivo:
            return camino

        if nodo_actual not in visitados:
            visitados.add(nodo_actual)

            for vecino in grafo[nodo_actual]:
                # COMPLETAR: crear un nuevo camino que sea "camino + [vecino]"
                # y agregarlo a la cola
                nuevo_camino = camino + [vecino]
                cola.append(nuevo_camino)

    return None # no se encontro un camino

resultado = bfs(mapa, "Almacen", "ZonaNorte")
print("Camino encontrado:", resultado)
```

## 4.4. Vectores y matrices con NumPy

Cada observación contiene dos características:

- Temperatura.
- Humedad.

Las observaciones serán agrupadas dentro de una matriz denominada `dataset`.

```python
import numpy as np

obs1 = np.array([22.5, 60])
obs2 = np.array([19.0, 45])
obs3 = np.array([25.3, 70])
obs4 = np.array([21.0, 50])
dataset = np.array([obs1, obs2, obs3, obs4])

print(dataset)
print("Forma de la matriz (filas, columnas):", dataset.shape)

print("Primera observacion (fila 0):", dataset[0])
print("Todas las temperaturas (columna 0):", dataset[:, 0])
```

## 4.5. Producto escalar y distancia euclidiana

Se calculará el producto escalar entre dos observaciones y posteriormente la distancia euclidiana de forma manual y mediante NumPy.

```python
producto = np.dot(obs1, obs2)

print("Producto escalar entre obs1 y obs2:", producto)
distancia_manual = np.sqrt(np.sum((obs1 - obs2) ** 2))

print("Distancia manual:", distancia_manual)
distancia_numpy = np.linalg.norm(obs1 - obs2)

print("Distancia con NumPy:", distancia_numpy)
```

## 5.2. Implementación de DFS

DFS realiza una búsqueda en profundidad utilizando una pila LIFO.

A diferencia de BFS, DFS explora primero un camino hasta avanzar lo máximo posible antes de retroceder.

```python
def dfs(grafo, inicio, objetivo):
    """
    Busca un camino entre 'inicio' y 'objetivo' usando DFS.
    """

    pila = [[inicio]]
    visitados = set()

    while pila:
        camino = pila.pop()
        nodo_actual = camino[-1]

        if nodo_actual == objetivo:
            return camino

        if nodo_actual not in visitados:
            visitados.add(nodo_actual)

            for vecino in grafo[nodo_actual]:
                nuevo_camino = camino + [vecino]
                pila.append(nuevo_camino)

    return None
resultado_dfs = dfs(mapa, "Almacen", "ZonaNorte")

print("Camino encontrado (DFS):", resultado_dfs)
print("Camino BFS:", resultado)
print("Camino DFS:", resultado_dfs)
```

## 5.3. Encontrar la observación más cercana

Se utilizará una nueva observación:

`[20.5, 48]`

Se calculará la distancia euclidiana entre esta observación y cada elemento del dataset para determinar cuál es la más cercana.

```python
nueva_observacion = np.array([20.5, 48])
distancias = []

for i, obs in enumerate(dataset):
    d = np.linalg.norm(nueva_observacion - obs)

    distancias.append(d)

    print(f"Distancia a la observacion {i}: {d}")
indice_mas_cercano = np.argmin(distancias)

print("La observacion mas parecida es la numero", indice_mas_cercano)
print("Observacion mas cercana:", dataset[indice_mas_cercano])
```

## 5.4. Reflexión

La distancia euclidiana es una forma razonable de medir qué tan parecidas son dos observaciones porque calcula la distancia existente entre sus características.

Mientras menor sea la distancia, mayor será la similitud entre las observaciones.

Sin embargo, puede presentar problemas cuando las características tienen escalas muy diferentes. Por ejemplo, una variable podría estar expresada en soles y otra en años.

En estos casos, una característica con valores mucho mayores puede influir demasiado en el resultado.

Por ello, puede ser necesario normalizar o estandarizar los datos antes de calcular las distancias.
